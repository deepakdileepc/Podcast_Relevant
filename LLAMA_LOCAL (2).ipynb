```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM, BitsAndBytesConfig
```

```python
!huggingface-cli login
```

```python
# Define quantization configuration
quantization_config = BitsAndBytesConfig(
    load_in_4bit=True,           # Use 4-bit quantization
    bnb_4bit_compute_dtype=torch.float16,  # Use FP16 for efficiency
    bnb_4bit_use_double_quant=True,  # Enable double quantization
    bnb_4bit_quant_type="nf4"   # Use NF4 (Normalized Float 4)
)
```

```python

# Load tokenizer for Llama-3.1-8B-Instruct
tokenizer = AutoTokenizer.from_pretrained("meta-llama/Llama-3.1-8B-Instruct")

# Load model in 4-bit precision for better performance on GPU
model = AutoModelForCausalLM.from_pretrained(
    "meta-llama/Llama-3.1-8B-Instruct",
    quantization_config=quantization_config,
    device_map="auto"           # Automatically use GPU
)
```

```python
# Function to classify text
def classify_text_local(input_text, video_summary):
    """Classify text locally using Llama-3.1-8B-Instruct"""
    system_message = "You are an AI assistant that helps classify text segments from YouTube video transcripts."

    user_message = f'''Task: Classify the following text segment as CASUAL or NOT CASUAL based on its relevance to the video's main topic and its informational value.

Guidelines:
CASUAL: The segment is entirely unrelated to the video's main topic. It provides no meaningful information or value to the video's primary subject. Examples: Off-topic remarks, personal anecdotes, jokes, small talk, or unrelated commentary.

NOT CASUAL: The segment is directly related to the video's main topic. It provides meaningful information, insights, or value to the video's primary subject. Examples: Explanations, instructions, discussions, or commentary aligned with the video's topic.

Strict Rule: Mark a segment as CASUAL only if it is completely unrelated or adds no value to the video's main topic. Even if the tone is informal, classify it as NOT CASUAL if it contributes meaningfully to the video's subject.

VIDEO SUMMARY: {video_summary}
TEXT TO CLASSIFY: {input_text}

Respond with exactly two lines:
- Classification: (CASUAL / NOT CASUAL)
- Explanation: Provide a brief reasoning explaining why'''

    # Format the prompt according to Llama's chat template
    messages = [
        {"role": "system", "content": system_message},
        {"role": "user", "content": user_message}
    ]

    # Apply the chat template to format the messages
    prompt = tokenizer.apply_chat_template(messages, tokenize=False)

    # Tokenize input
    inputs = tokenizer(prompt, return_tensors="pt").to("cuda")  # Move to GPU

    # Generate response
    outputs = model.generate(**inputs, max_new_tokens=150, temperature=0.7)

    # Decode response
    generated_text = tokenizer.decode(outputs[0][inputs.input_ids.shape[1]:], skip_special_tokens=True)

    # Extract classification and explanation
    classification = None
    explanation = None

    if "Classification:" in generated_text:
        classification_start = generated_text.find("Classification:") + len("Classification:")
        classification_end = generated_text.find("Explanation:")
        if classification_end > classification_start:
            classification_text = generated_text[classification_start:classification_end].strip()
            if "NOT CASUAL" in classification_text:
                classification = "NOT CASUAL"
            elif "CASUAL" in classification_text:
                classification = "CASUAL"

            if "Explanation:" in generated_text:
                explanation_start = generated_text.find("Explanation:") + len("Explanation:")
                explanation = generated_text[explanation_start:].strip()

    return {
        'input_text': input_text,
        'classification': classification,
        'explanation': explanation
    }
```

```python
import yaml

# Load the transcript
with open('input.yaml', 'r') as file:
    data = yaml.safe_load(file)

# Extract input texts
input_texts = [d['text'] for d in data]

video_summary="he video chronicles a 24-hour culinary journey through various Gordon Ramsay restaurants. The creator begins at The Savoy Grill with a full English breakfast, appreciating the sausage and eggs but finding the hash brown unremarkable. Next, he visits Street Pizza, sampling a variety of bottomless pizzas, with the ham and pineapple emerging as his favorite. At Street Burger, he tries the Hell's Kitchen Burger, commending its composition and relish, though the fries are considered average. Lucky Cat offers a three-course lunch, featuring sushi, flat iron steak with salmon teriyaki, and a chocolate tofu with mochi dessert, earning a high rating of 8.5 out of 10. The journey concludes back at The Savoy Grill with a massive and delicious beef Wellington and a sticky toffee pudding, which receives a 9 out of 10."
# Run classification sequentially
results = [classify_text_local(text, video_summary) for text in input_texts]

# Save results to a pickle file
import pickle
with open("results.pkl", 'wb') as file:
    pickle.dump(results, file)

# Reload and display a few results
with open("results.pkl", 'rb') as file:
    results = pickle.load(file)

# Display the first two results
for result in results[:2]:
    print(result)
```
